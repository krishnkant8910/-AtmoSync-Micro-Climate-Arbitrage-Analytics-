```python
import pandas as pd

# Load the dataset using the GitHub raw file link
url = "https://raw.githubusercontent.com/krishnkant8910/-AtmoSync-Micro-Climate-Arbitrage-Analytics-/refs/heads/main/Indian_Climate_Dataset_2024_2025.csv"

df = pd.read_csv(url)

# Display the first 5 rows of the dataset
print(df.head())
```

```python
# Display the column names available in the dataset
print(df.columns)
```

```python
# 1. Check for any missing (null) values in the dataset
print("--- Missing Values ---")
print(df.isnull().sum())

# 2. Display the basic statistical summary (Mean, Min, Max, etc.) for temperature and humidity
print("\n--- Statistical Summary ---")
print(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a boxplot to visualize the temperature distribution across cities
plt.figure(figsize=(12, 6))
sns.boxplot(
    x="City", y="Temperature_Max (°C)", data=df
)  # (Update the column names if they differ in your dataset)
plt.xticks(rotation=45)
plt.title("Temperature Distribution Across Cities")
plt.show()
```

```python
# Group by city to find the maximum temperature and create a sorted list of the hottest cities
top_hot_cities = (
    df.groupby("City")["Temperature_Max (°C)"].max().reset_index()
)
top_hot_cities = top_hot_cities.sort_values(
    by="Temperature_Max (°C)", ascending=False
)

print("--- Top 10 Hottest Cities ---")
print(top_hot_cities.head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Extract the data for the top 10 hottest cities
top_10 = top_hot_cities.head(10)

# Plot a bar chart for the top 10 hottest cities
plt.figure(figsize=(12, 6))
sns.barplot(x="City", y="Temperature_Max (°C)", data=top_10, palette="Reds_r")
plt.xticks(rotation=45)
plt.title("Top 10 Hottest Cities by Maximum Temperature")
plt.xlabel("City")
plt.ylabel("Max Temperature (°C)")
plt.show()
```

```python
# Calculate and display the average for all numerical columns in the dataset
print(df.mean(numeric_only=True))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Compute and plot the correlation matrix for numerical climate factors
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix of Climate Factors")
plt.show()
```

```python
# Print the overall summary of the AtmoSync project metrics
print("=== ATMOSYNC PROJECT SUMMARY ===")
print(f"Total Cities Analyzed: {df['City'].nunique()}")
print(
    f"Highest Recorded Temperature: {df['Temperature_Max (°C)'].max():.2f} °C"
)
print(f"Average Dataset Temperature: {df['Temperature_Max (°C)'].mean():.2f} °C")
```
